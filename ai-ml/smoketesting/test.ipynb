```python
import joblib
from pathlib import Path

MODEL_PATH = Path('../../models/core-model/xgb/checkpoints/optimised_core_xgb_20260926_184719.joblib')  # <-- set this  

bundle = joblib.load(MODEL_PATH)
print(type(bundle))
print("Model expects:", bundle["model"].n_features_in_, "features")

print(list(bundle.keys()) if isinstance(bundle, dict) else "not a dict")
for k, v in bundle.items():
    print(f"  {k}: {type(v)}")

CHECKPOINT_DIR = Path('../../models/core-model/xgb/checkpoints')

candidates = list(CHECKPOINT_DIR.glob("tfidf_*")) + \
             list(CHECKPOINT_DIR.glob("*vector*")) + \
             list(CHECKPOINT_DIR.glob("*.pkl"))

for f in sorted(candidates, key=lambda p: p.stat().st_mtime, reverse=True):
    print(f.name, "-- modified:", f.stat().st_mtime)
    
text_vec = joblib.load("../../models/core-model/xgb/checkpoints/tfidf_text_20260926_184719.pkl") 
print("Vectorizer vocab size:", len(text_vec.get_feature_names_out()))
print("Model expects:", bundle["model"].n_features_in_, "total") 
print("Implied numeric feature count:", bundle["model"].n_features_in_ - len(text_vec.get_feature_names_out()))



model_n_features = bundle["model"].n_features_in_
numeric_features = bundle["numeric_features"]
text_vectorizers = bundle["text_vectorizers"]

vocab_sizes = {name: len(vec.get_feature_names_out()) for name, vec in text_vectorizers.items()}
total_text_vocab = sum(vocab_sizes.values())

print("Numeric features:", len(numeric_features), numeric_features)
print("Text vectorizers:", list(text_vectorizers.keys()))
print("Vocab sizes:", vocab_sizes)
print("Total expected:", len(numeric_features) + total_text_vocab)
print("Model actually expects:", model_n_features)

assert len(numeric_features) + total_text_vocab == model_n_features, "Mismatched, do not proceed"
print("✅ Bundle is internally consistent")
```
